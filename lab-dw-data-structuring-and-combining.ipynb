```python
customer_data_insurance_df = pd.read_csv("Customer_data.csv")
customer_data_insurance2_df = pd.read_csv("Customer_data2.csv")
customer_data_insurance3_df = pd.read_csv("Customer_data3.csv")
```

```python
all_customer_data_df = pd.concat([customer_data_insurance_df, customer_data_insurance2_df, customer_data_insurance3_df], axis=0, ignore_index=True)
```

```python
all_customer_data_df.info()
```

```python
all_customer_data_df["state"] = all_customer_data_df["ST"].fillna(all_customer_data_df["State"])
```

```python
all_customer_data_df = all_customer_data_df.drop(["ST", "State"], axis=1)
```

```python
all_customer_data_df.info()
```

```python
all_customer_data_df["gender"] = all_customer_data_df["GENDER"].fillna(all_customer_data_df["Gender"])
all_customer_data_df = all_customer_data_df.drop(["GENDER", "Gender"], axis=1)
```

```python
all_customer_data_df.info()
```

```python
def cle_for(df, state_conversion, gender_conversion):
    # Cleaning columns
    df.columns = df.columns.str.lower().str.replace(' ', '_')

    # Drop rows where all elements are NaN
    df = df.dropna(axis=0, how="all")
    
    # Map and replace categories
    df["state"] = df["state"].map(state_conversion).fillna(df["state"])
    df["gender"] = df["gender"].map(gender_conversion).fillna(df["gender"])
    return df
```

```python
state_conversion = {
    'Cali': 'California', 'AZ': 'Arizona', 'WA': 'Washington',
    'Washington': 'Washington', 'Arizona': 'Arizona', 'Nevada': 'Nevada', 'California': 'California', 'Oregon': 'Oregon'
}
```

```python
gender_conversion = {
    "Male": "M", "Female": "F", "Femal": "F", "female": "F"
}
```

```python
cleaned_customer_data_insurance_df = cle_for(all_customer_data_df, state_conversion, gender_conversion)
cleaned_customer_data_insurance_df = cleaned_customer_data_insurance_df.reset_index(drop=True)
```

```python
cleaned_customer_data_insurance_df.columns
```

```python
cleaned_customer_data_insurance_df["education"].unique()
```

```python
cleaned_customer_data_insurance_df["education"] = cleaned_customer_data_insurance_df["education"].replace({
    "Bachelors": "Bachelor"
})
```

```python
cleaned_customer_data_insurance_df["education"].unique()
```

```python
cleaned_customer_data_insurance_df["customer_lifetime_value"].unique()
```

```python
cleaned_customer_data_insurance_df["customer_lifetime_value"] = cleaned_customer_data_insurance_df["customer_lifetime_value"].str.replace('%', '', regex=False)
cleaned_customer_data_insurance_df["customer_lifetime_value"] = pd.to_numeric(
    cleaned_customer_data_insurance_df["customer_lifetime_value"], errors='coerce'
    )
median_value = cleaned_customer_data_insurance_df["customer_lifetime_value"].median()
cleaned_customer_data_insurance_df["customer_lifetime_value"] = cleaned_customer_data_insurance_df["customer_lifetime_value"].fillna(median_value)
```

```python
cleaned_customer_data_insurance_df["number_of_open_complaints"] = cleaned_customer_data_insurance_df["number_of_open_complaints"].astype(str)
```

```python
cleaned_customer_data_insurance_df["number_of_open_complaints"].dtypes
```

```python
def convert_values(x):
    if pd.isnull(x):
        return pd.NA  # Handle missing values
    if isinstance(x, str) and "/" in x:
        return int(x.split("/")[1])  # Extract the second part of the split string
    try:
        return int(x)  # Directly convert numeric strings or numbers
    except ValueError:
        return pd.NA  # In case of any unexpected values

cleaned_customer_data_insurance_df["number_of_open_complaints"] = (
    cleaned_customer_data_insurance_df["number_of_open_complaints"]
    .apply(convert_values)
    .astype(pd.Int64Dtype())
)
```

1. You work at the marketing department and you want to know which sales channel brought the most sales in terms of total revenue. Using pivot, create a summary table showing the total revenue for each sales channel (branch, call center, web, and mail).
Round the total revenue to 2 decimal points.  Analyze the resulting table to draw insights.

```python
premium_pivot_table = marketing_customer_analysis_df.pivot_table(
    values='monthly_premium_auto',
    index='sales_channel',
    aggfunc='sum'
)

premium_pivot_table = premium_pivot_table.round(2)

print(premium_pivot_table)
```

```python
#The data might hint at market preferences that skew toward personal interaction.
# This could be critical information when planning marketing campaigns or customer interaction strategies.
```

2. Create a pivot table that shows the average customer lifetime value per gender and education level. Analyze the resulting table to draw insights.

```python
pivot_table_clv = marketing_customer_analysis_df.pivot_table(
    values='customer_lifetime_value',
    index='gender',
    columns='education',
    aggfunc='mean'
)

pivot_table_clv = pivot_table_clv.round(2)

print(pivot_table_clv)
```

```python
#Given the higher CLVs among females with "High School or Below" and males with "Masters," these groups could be prime targets for retention strategies and upselling higher insurance packages or additional service offerings.
```

## Bonus

You work at the customer service department and you want to know which months had the highest number of complaints by policy type category. Create a summary table showing the number of complaints by policy type and month.
Show it in a long format table.

*In data analysis, a long format table is a way of structuring data in which each observation or measurement is stored in a separate row of the table. The key characteristic of a long format table is that each column represents a single variable, and each row represents a single observation of that variable.*

*More information about long and wide format tables here: https://www.statology.org/long-vs-wide-data/*

```python
df_complaints = marketing_customer_analysis_df.groupby(['policy_type', 'month']).size().reset_index(name='number_of_complaints')

print(df_complaints)
```
