```python
import numpy as np
import pandas as pd
```

```python
marketing_customer_analysis_df = pd.read_csv("marketing_customer_analysis.csv")
```

```python
marketing_customer_analysis_df.nunique()
```

```python
marketing_customer_analysis_df.info()
```

```python
marketing_customer_analysis_df.columns = marketing_customer_analysis_df.columns.str.lower().str.replace(' ', '_')
```

```python
marketing_customer_analysis_df.columns
```

```python
#1. Creating a new dataframe 
filtered_df = marketing_customer_analysis_df[
    (marketing_customer_analysis_df["total_claim_amount"] < 1000) &  # Condition for low claim amount
    (marketing_customer_analysis_df["response"] == "Yes")           # Condition for positive response
]

# Display the filtered DataFrame
print(filtered_df)
```

```python
#2.
# Step 1: Filter the DataFrame for 'Yes' responses
responded_yes_df = marketing_customer_analysis_df[marketing_customer_analysis_df["response"] == "Yes"]

# Step 2: Calculate averages
average_metrics = responded_yes_df.groupby(["policy_type", "gender"]).agg({
    "monthly_premium_auto": "mean",
    "customer_lifetime_value": "mean"
}).round(2)

# Step 3: Analyze `total_claim_amount`
average_claims = responded_yes_df.groupby(["policy_type", "gender"])["total_claim_amount"].mean().round(2)

# Display results
print("Average Metrics:")
print(average_metrics)

print("\nAverage Total Claim Amount:")
print(average_claims)
```

```python
#3
# Group by 'state' and count the number of customers in each state
customers_per_state = marketing_customer_analysis_df.groupby("state").size().reset_index(name="customer_count")

# Filter the states to include only those with more than 500 customers
states_with_large_customer_base = customers_per_state[customers_per_state["customer_count"] > 500]

print(states_with_large_customer_base)
```

```python
#4
# Group by 'education' and 'gender' and calculate the required statistics
clv_stats_by_group = marketing_customer_analysis_df.groupby(["education", "gender"])["customer_lifetime_value"].agg(["max", "min", "median"])

# Display the results
print(clv_stats_by_group)
```

```python
#Conclusions:
#Higher CLV in males with lower education levels could suggest opportunities for engagement initiatives targeting this demographic with educational content, 
# up-sell opportunities, or loyalty programs.
#Differences in median and maximum values between genders can guide gender-focused campaign elements, 
#ensuring communication is tailored to engage effectively with each group.
```

```python
#5
# Convert the 'Effective To Date' column to datetime
marketing_customer_analysis_df["effective_to_date"] = pd.to_datetime(marketing_customer_analysis_df["effective_to_date"])

# Extract the month name for each entry
marketing_customer_analysis_df["month"] = marketing_customer_analysis_df["effective_to_date"].dt.month_name()

# Create a pivot table
policies_by_state_and_month = marketing_customer_analysis_df.pivot_table(
    index="state",
    columns="month",
    values="policy",  
    aggfunc="count"
)

# Reorder the columns to maintain calendar order
policies_by_state_and_month = policies_by_state_and_month.reindex(columns=[
    "January", "February", "March", "April", "May", "June", 
    "July", "August", "September", "October", "November", "December"
])

# Display the pivot table
print(policies_by_state_and_month)
```

```python
#6
# Group by state and month to count the number of policies sold
policies_by_state_month = marketing_customer_analysis_df.groupby(["state", "month"]).size().reset_index(name="policies_sold")

# Calculate total policies sold by state
total_policies_by_state = policies_by_state_month.groupby("state")["policies_sold"].sum().reset_index()

# Sort and get the top 3 states with the highest number of policies sold
top_3_states = total_policies_by_state.sort_values(by="policies_sold", ascending=False).head(3)["state"]

# Filter for the top 3 states
top_3_policies_by_month = policies_by_state_month[policies_by_state_month["state"].isin(top_3_states)]

# Optional: reorder the months to ensure they appear in calendar order
calendar_months = ["January", "February", "March", "April", "May", "June", 
                   "July", "August", "September", "October", "November", "December"]
top_3_policies_by_month['Month'] = pd.Categorical(top_3_policies_by_month["month"],
                                                  categories=calendar_months,
                                                  ordered=True)

# Display the resulting DataFrame
print(top_3_policies_by_month.sort_values(by=["state", "month"]))
```

```python
#7
# Filter and calculate the number of "Yes" responses per channel
yes_responses = marketing_customer_analysis_df[marketing_customer_analysis_df["response"] == "Yes"].groupby("sales_channel").size().reset_index(name="yes_count")

# Calculate total responses per channel
total_responses = marketing_customer_analysis_df.groupby("sales_channel").size().reset_index(name="total_count")

# Merge the dataframes based on "sales_channel"
response_data = pd.merge(yes_responses, total_responses, on="sales_channel")

# Calculate the response rate
response_data["response_rate"] = (response_data["yes_count"] / response_data["total_count"]) * 100

# Display the resulting DataFrame
print(response_data)
```
